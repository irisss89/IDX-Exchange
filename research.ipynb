```python
bay_area_counties = [
    'Santa Clara', 'San Mateo', 'San Francisco',
    'Alameda', 'Contra Costa', 'Marin', 'Napa', 'Sonoma', 'Solano'
]

df_bay = sold_combine[sold_combine['CountyOrParish'].isin(bay_area_counties)]
df_bay = df_bay[df_bay['StateOrProvince'] == 'CA']
print(df_bay.head(5))
```

```python
# Convert dates
df_bay['CloseDate'] = pd.to_datetime(df_bay['CloseDate'], errors='coerce')

# Remove invalid values
df_bay = df_bay[
    (df_bay['ClosePrice'] > 0) &
    (df_bay['LivingArea'] > 0) &
    (df_bay['BedroomsTotal'] >= 0) &
    (df_bay['BathroomsTotalInteger'] >= 0)
]

# Drop missing key fields
df_bay = df_bay.dropna(subset=['ClosePrice', 'LivingArea', 'CloseDate'])
```

```python
# Price per sqft
df_bay['price_per_sqft'] = df_bay['ClosePrice'] / df_bay['LivingArea']

# Year / Month
df_bay['year'] = df_bay['CloseDate'].dt.year
df_bay['month'] = df_bay['CloseDate'].dt.to_period('M')

# Age of house
df_bay['house_age'] = df_bay['CloseDate'].dt.year - df_bay['YearBuilt']

# Listing vs selling difference
df_bay['price_diff'] = df_bay['ClosePrice'] - df_bay['ListPrice']
df_bay['price_ratio'] = df_bay['ClosePrice'] / df_bay['ListPrice']
```

```python
print(df_bay['ClosePrice'].describe())
print(df_bay['price_per_sqft'].describe())
```

```python
monthly_trend = df_bay.groupby('month')['ClosePrice'].median()

monthly_trend.plot(figsize=(10,5), title='Median Price Over Time')
```

```python
county_summary = df_bay.groupby('CountyOrParish').agg({
    'ClosePrice': 'median',
    'price_per_sqft': 'median',
    'DaysOnMarket': 'median'
}).sort_values('ClosePrice', ascending=False)

print(county_summary)
```

```python
city_summary = df_bay.groupby('City')['ClosePrice'].median().sort_values(ascending=False)

print(city_summary.head(20))
```

```python
ptype_summary = df_bay.groupby('PropertySubType')['ClosePrice'].median()

print(ptype_summary)
```

```python
zip_summary = df_bay.groupby('PostalCode')['ClosePrice'].median().sort_values(ascending=False)
print(zip_summary.head(20))
```

```python
yearly_trend = df_bay.groupby('year')['ClosePrice'].median()
print(yearly_trend)
```

```python
growth = (yearly_trend.loc[2026] / yearly_trend.loc[2024] - 1) * 100
print(f"Price change from 2024 to 2026: {growth:.2f}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(monthly_trend.index.astype(str), monthly_trend.values)
plt.xticks(rotation=45)
plt.title('Median Price Over Time (Bay Area)')
plt.ylabel('Price')
plt.show()
```
